### Functions

```python
def movimiento_browniano(delta_t, T, semilla):
    n = int(T / delta_t)
    np.random.seed(semilla)
    X = np.random.choice([-1, 1], size=(n, 3))
    W = np.zeros((n, 3))
    for i in range(1, n):
        W[i] = W[i-1] + X[i] * np.sqrt(delta_t)
    fig = go.Figure()
    fig.add_trace(go.Scatter3d(x=W[:, 0], y=W[:, 1], z=W[:, 2], mode="lines", name="Trayectoria"))
    fig.add_trace(go.Scatter3d(x=[W[0, 0]], y=[W[0, 1]], z=[W[0, 2]], mode="markers", name="Inicio", marker=dict(size=8, color="green")))
    fig.add_trace(go.Scatter3d(x=[W[-1, 0]], y=[W[-1, 1]], z=[W[-1, 2]], mode="markers", name="Fin", marker=dict(size=8, color="red")))
    fig.update_layout(title="Movimiento Browniano 3D", scene=dict(xaxis_title="X", yaxis_title="Y", zaxis_title="Z"))
    fig.show()

def apuesta_individual(capital_inicial, semilla):
    np.random.seed(semilla)
    capital = capital_inicial
    capitales = [capital]
    while capital > 0:
        apuesta = capital / 2 if capital >= 6 else 2
        resultado = np.random.choice([-1, 1])
        capital = max(capital + resultado * apuesta, 0)
        capitales.append(capital)
    fig = go.Figure()
    fig.add_trace(go.Scatter(x=np.arange(len(capitales)), y=capitales, mode="lines+markers", name="Capital"))
    fig.add_hline(y=6, line_dash="dash", annotation_text="Umbral de $6")
    fig.update_layout(title="Evolución del capital", xaxis_title="Número de volados", yaxis_title="Capital (pesos)")
    fig.show()

def monte_carlo_apuestas(capital_inicial, n, semilla):
    np.random.seed(semilla)
    volados = []
    for _ in range(n):
        capital = capital_inicial
        pasos = 0
        while capital > 0:
            apuesta = capital / 2 if capital >= 6 else 2
            resultado = np.random.choice([-1, 1])
            capital = max(capital + resultado * apuesta, 0)
            pasos += 1
        volados.append(pasos)
    volados = np.array(volados)
    fig = go.Figure()
    fig.add_trace(go.Histogram(x=volados, nbinsx=50, name="Volados"))
    fig.add_vline(x=volados.mean(), line_dash="dash", annotation_text=f"Promedio: {volados.mean():.2f}")
    fig.update_layout(title="Número de volados hasta la quiebra", xaxis_title="Número de volados", yaxis_title="Frecuencia")
    fig.show()
```

### Movimiento Browniano 

### Fórmula Recursiva



El movimiento browniano es el movimiento aleatorio que se observa en las partículas que se hallan en un medio fluido, como resultado de choques contra las moléculas de dicho fluido.

Para simular el movimiento, en lugar de usar las ecuaciones integro-diferenciales que describe Einstein, use una caminata aleatoria para los tres ejes $x$, $y$, $z$ de la siguiente forma

$$
W_i = W_{i-1} + X_i\sqrt{\Delta t}
$$

donde, $\Delta t$ es un intervalo de tiempo y

$$
X_i =
\begin{cases}
1 \\
-1
\end{cases}
$$

se escoge de forma aleatoria con la misma probabilidad.

Simule 10 segundos con un $\Delta t = 0.001$ y represéntelo en un gráfico 3D.


$$n = \frac{10}{0.001} = 10,000$$

### Apuesta Arriesgada 


Un jugador, con $20$ pesos de capital inicial, decide apostar a un volado (con una moneda perfectamente nivelada) la mitad de su capital si éste es mayor o igual a $6$ pesos; de lo contrario, solo $2$ pesos. Esto es, el capital $C_i$ en cada volado $V_i$ se describe como:

### Fórmula Recursiva


$$
C_{i+1} = C_i + V_i A_i
$$

donde la apuesta es:

$$
A_i =
\begin{cases}
C_i/2, & \text{si } C_i \geq 6 \\
2, & \text{de lo contrario}
\end{cases}
$$
