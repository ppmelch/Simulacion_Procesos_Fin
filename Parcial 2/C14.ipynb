#### Libraries 

```python
import numpy as np
import pandas as pd 
import seaborn as sns
from scipy.stats import qmc
from scipy.stats import norm
import matplotlib.pyplot as plt

from warnings import filterwarnings
filterwarnings('ignore')
```

---

#### Functions

```python
def f(x):
    return x**2

def simular(m):
    exacto = 1/3

    x_mc = np.random.default_rng(20).uniform(0, 1, m)
    x_qmc = qmc.Sobol(d=1, scramble=True, seed=20).random(m).flatten()

    est_mc = np.mean(f(x_mc))
    est_qmc = np.mean(f(x_qmc))

    return [m, est_mc, est_qmc,
            abs(est_mc - exacto), abs(est_qmc - exacto)]


def simular_probabilidad(m):
    u_mc = np.random.default_rng(20).uniform(0, 1, m)
    u_qmc = qmc.Sobol(d=1, scramble=True, seed=20).random(m).flatten()

    p_mc = np.mean(norm.ppf(u_mc, 50, 10) > 60)
    p_qmc = np.mean(norm.ppf(u_qmc, 50, 10) > 60)

    return [m, p_mc, p_qmc]
```

---

```python
m_valores = [100, 500, 1000, 5000]

tabla = pd.DataFrame(
    [simular(m) for m in m_valores],
    columns=["m", "Estimación MC", "Estimación QMC", "Error MC", "Error QMC"]
)

display(tabla)
```

```python
tabla2 = pd.DataFrame(
    [simular_probabilidad(m) for m in [10, 100, 1000, 10000]],
    columns=["m", "MC", "QMC"]
)
display(tabla2)
```

---

#### Conclusión 

Con estoo se pudo ver que los dos métodos sirven para aproximar resultados, pero QMC obtuvo valores más cercanos a los resultados esperados. También se observa que, al aumentar la cantidad de simulaciones, las estimaciones mejoran. En general, QMC tuvo un mejor resultado en las pruebas realizadas.
