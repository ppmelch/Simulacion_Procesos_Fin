```python
import numpy as np
seed = np.random.seed(32)
```

```python
# Simulación dado con simulación de 10-100,000 datos 

sim = [10 , 100, 1000, 10000, 100000]

def datos(n):
    data = np.random.randint(1,6, size=n)
    return data

for i in sim:
    muestra = datos(i)
    E = np.mean(muestra)
    print(i, E)    
```
