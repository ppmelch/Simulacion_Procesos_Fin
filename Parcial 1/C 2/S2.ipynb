### Formúla de Números pseudoaleatorios


$$X_{n+1} = (a \cdot X_n + b) \bmod m$$

---

10 numeros pseudoaleatoirios de la formula 

```python

m = 9
a = 5 
b = 1 
x0 = 21
x = [x0]

for i in range(1, 11):
    X = (a * x[i-1] + b) % m
    x.append(X)
resultados = x[1:]
print(resultados)
```

```python
m = 9
a = 5 
b = 1 
x0 = 21
x = [x0]

for i in range(1, 11):
    X = (b * x[i-1] + a) % m
    x.append(X)
resultados = x[1:]
print(resultados)
```

### Método Box-Muller

Generar dos cadenas uniforme que asemejen 2 

$$N(0,1)$$
 
$$U_1 \text{ y } U_2 \text{ ∈} \text{ U(0,.9)}$$


$$X = R\cos\theta = \sqrt{-2\ln(U_1)}\,\cos(2\pi U_2)$$

$$Y = R\sin\theta = \sqrt{-2\ln(U_1)}\,\sin(2\pi U_2)$$



$$R = \sqrt{-2ln(U_1)}$$

$$\theta = 2\pi U_2$$

```python
m = 16
a = 5
b = 3
x01 = 0
x02 = 3
n = 30
valor = [lim_izq + (lim_der - lim_izq) * (i / m) for i in resultados]

x1 = [x01]
for i in range(n):
    x1.append((a * x1[-1] + b) % m)
resultados1 = x1[1:]

x2 = [x02]
for i in range(n):
    x2.append((a * x2[-1] + b) % m)
resultados2 = x2[1:]

lim_izq, lim_der = 0, 0.9

U1 = np.array([lim_izq + (lim_der - lim_izq) * (i / m) for i in resultados1])
U2 = np.array([lim_izq + (lim_der - lim_izq) * (i / m) for i in resultados2])

R = np.sqrt(-2 * np.log(U1 + 1e-10))   
theta = 2 * np.pi * U2

X = R * np.cos(theta)
Y = R * np.sin(theta)

plt.hist(X, bins=10, density=True, color='blue', alpha=0.7, edgecolor='black')
plt.show()
```
