```python
def derivada_discreta(V_k1 , Vk , S_k1, S_k):
    if S_k1 - S_k == 0:
        raise ValueError("La diferencia entre S_k1 y S_k no puede ser cero.")
    return (V_k1 - Vk) / (S_k1 - S_k)



    
```
